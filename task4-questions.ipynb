### Question 1 

What is the effect of increasing the number of layers in an MLP? Based on your experiments in task 2/3, are larger models (with a greater number of hidden layers) preferred to smaller ones? Why or why not?

   Your answer: **Adding more hidden layers gives the network more capacity. Each layer can build on the features found by the one before it, so a deeper MLP can in principle learn more complex patterns. But bigger is not automatically better. More layers means many more weights, so the model is slower to train, needs more epochs and careful tuning of the learning rate and regularization, and is more likely to overfit (training accuracy keeps going up while validation accuracy stops improving). Deeper networks are also harder to optimize because the gradients have to pass back through more layers and can shrink or blow up along the way. In my experiments, the biggest improvements came from training choices like using momentum and tuning the learning rate and regularization, rather than from simply stacking more layers. Also, since an MLP flattens the image, extra layers still can't use the spatial structure of the tissue images. So larger models are not automatically preferred: the best choice is the smallest model that reaches good validation accuracy, and you only add layers if validation accuracy actually improves.**

### Question 2 

What is the significance of activation functions in deep learning models? Name two activation functions that you can use for a hidden layer **and** two for the output layer. 

   Your answer: **Activation functions add non-linearity to the network. Without them, stacking layers would be pointless, because a chain of linear layers is still just one linear function, so the network could only learn linear decision boundaries no matter how deep it is. With a non-linear activation between layers, the network can learn complex, curved relationships in the data. The choice of activation also affects how well the gradients flow during training (for example, ReLU avoids the vanishing gradients that sigmoid and tanh can have). Two activations for a hidden layer: ReLU (or leaky ReLU) and tanh. Two activations for the output layer: softmax (for multi-class classification, it turns the scores into probabilities that sum to 1) and sigmoid (for binary classification, it gives a probability between 0 and 1). For a regression output you could also use a linear (identity) activation.**

### Question 3

Assume you have a problem to predict the annual rainfall in a certain region with some historic numeric data that was given to you, which of these 2 models (Linear Regression vs. Logistic Regression) would you use? How would you modify the problem statement to use the other model? Do they both adopt a linear decison boundary? 

   Your answer: **Rainfall is a continuous number, so I would use Linear Regression, which predicts a real-valued output. Logistic Regression predicts the probability of a class, so it is meant for classification. To use Logistic Regression, I would turn the problem into a classification task, for example: "Will the annual rainfall be above a certain threshold (e.g. above the historical average) this year: yes or no?" The labels then become 0/1, and the model predicts the probability of a high-rainfall year. You could also split rainfall into several ranges (low / medium / high) and use multi-class (softmax) logistic regression. As for decision boundaries: Logistic Regression does have a linear decision boundary, because it predicts class 1 when w^T x > 0, and that boundary is a straight line (a hyperplane) in the input space. Linear Regression doesn't really have a decision boundary since it isn't classifying anything. It fits a linear function (a line or hyperplane) to predict the value. So both are linear models, but only Logistic Regression uses a linear boundary to separate classes.**


### Question 4

What will happen if you choose a very small or a very large learning rate? 

   Your answer: **If the learning rate is too small, each update only moves the weights a tiny amount, so training is very slow. The loss decreases very gradually, you need many more epochs to get a good model, and training may stop (or the time budget runs out) before the model gets close to a good solution. It can also get stuck in flat regions or poor local minima. If the learning rate is too large, the updates overshoot the minimum. The loss jumps around instead of going down smoothly, the model may bounce back and forth without settling, and in the worst case the loss blows up (diverges) and the weights become huge or NaN. A good learning rate is somewhere in between, and it is common to start fairly large and then decay it over time (as our training function does every epoch) so the model makes fast progress early and then settles down.**

What is the interpretation of **perplexity** in t-SNE? How did you set this value during the tuning in task3?
    
   Your answer: **Perplexity can be thought of as roughly how many neighbors each point pays attention to when t-SNE builds the 2D map. A low perplexity makes t-SNE focus on very local structure (only a few nearest neighbors), which often breaks the data into many small, scattered clumps. A high perplexity makes each point consider many more neighbors, so the plot shows more of the global structure and gives larger, smoother clusters, but very high values can blur separate groups together. Typical values are between about 5 and 50, and it should be smaller than the number of points. In task 3, the given visualization used a perplexity of 25. For tuning, I tried a low value (5) and a high value (50) side by side, keeping the other settings the same (and using PCA initialization and a stronger early exaggeration), to compare how local versus global the resulting clusters look.**
